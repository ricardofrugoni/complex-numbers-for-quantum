# 02 — Módulo e argumento

Para $z=a+bi$, visualize $|z|$ como distância e $\arg(z)$ como ângulo.


```python
import cmath
import numpy as np

z = 3 + 4j
r, theta = cmath.polar(z)
print('z =', z)
print('|z| =', r)
print('arg(z) rad =', theta)
print('arg(z) graus =', np.degrees(theta))
```
