# Distância e direção contam a mesma história

$|z|=\sqrt{a^2+b^2}$ mede a distância à origem. Um argumento indica direção, para $z\ne0$. Compare as setas de mesmo comprimento. [Leia o capítulo](../docs/02_modulo_argumento.md).

```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / 'src' / 'complex_geometry.py').is_file()), None)
if ROOT is None:
    raise RuntimeError('Abra o notebook dentro da pasta do projeto.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
import cmath

fig, ax = plt.subplots(figsize=(6, 6))
for z in [1+1j, -1-1j, 1-1j]:
    ax.plot([0, z.real], [0, z.imag], 'o-', label=f'{z}: {np.degrees(cmath.phase(z)):.0f}°')
    print(f'{z}: módulo {abs(z):.4f}, argumento {np.degrees(cmath.phase(z)):.0f}°')
theta = np.linspace(0, 2*np.pi, 300)
ax.plot(np.sqrt(2)*np.cos(theta), np.sqrt(2)*np.sin(theta), ':', color='gray')
ax.axhline(0, color='gray', lw=.5); ax.axvline(0, color='gray', lw=.5)
ax.set(xlabel='Parte real', ylabel='Parte imaginária', aspect='equal')
ax.legend(); ax.grid(alpha=.2); plt.show()
```

## O ângulo pode saltar sem o ponto saltar

As direções 179° e 181° estão próximas. O argumento principal da segunda aparece como −179°. Na origem, o argumento matemático é indefinido, mesmo que a biblioteca forneça um valor por convenção.

```python
for degrees in [179, 181, 360]:
    z = cmath.rect(1, np.radians(degrees))
    print(f'Ângulo acumulado {degrees}° → principal {np.degrees(cmath.phase(z)):.1f}°')
z = 0j
argument = None if z == 0 else cmath.phase(z)
print('Argumento da origem:', 'indefinido' if argument is None else argument)
```
