## Uma rotação vista de lado

$z(t)=A e^{i(\omega t+\varphi)}$ tem projeção real $A\cos(\omega t+\varphi)$. Aqui $t$ está em segundos e $\omega$ em rad/s. Mova a amplitude, a fase e o instante: a posição real do ponto acompanha o sinal.

[Ondas e fasores](../docs/09_onde_sao_usados.md#ondas) · [Prévia estática](../assets/interactive/ondas.png).

```python
wave_lab = waves()
plt.show()
```

**Compare:** aumentar ω encurta o período; aumentar A muda a altura; alterar φ desloca o ciclo. Com A=0, o sinal é nulo e não há fase definida. Um fasor representa a amplitude complexa após separar a rotação comum.

## Duas frequências, muitas formas

$s(t)=\cos t+A\cos(kt+\varphi)$. O laboratório calcula a DFT de 512 amostras de uma volta completa. O harmônico inteiro mantém as frequências alinhadas à grade da DFT. O parâmetro $t$ deste exemplo é angular, em radianos.

[A janela para Fourier](../docs/09_onde_sao_usados.md#fourier) · [Prévia estática](../assets/interactive/fourier.png).

```python
fourier_lab = fourier_components()
plt.show()
```

**Observe:** a fase muda a forma do sinal sem mudar as alturas das barras. O coeficiente complexo da segunda componente muda. Cada cosseno divide sua amplitude entre as frequências $+k$ e $-k$, com módulo $A/2$ em cada coeficiente. O gráfico unilateral exibe $A$. Se $A=0$, a fase dessa componente é indefinida.

Este exemplo não ensina toda a análise de Fourier: sinais gerais também exigem cuidados com amostragem, aliasing e janelamento. [NumPy — DFT](https://numpy.org/doc/stable/reference/routines.fft.html).

## Uma janela para sistemas dinâmicos

$e^{(\sigma+i\omega)t}=e^{\sigma t}e^{i\omega t}$ combina variação de magnitude e rotação. A figura estática abaixo compara três valores de $\sigma$ com $\omega=1$: decaimento, círculo e crescimento. Equações diferenciais são o próximo passo para associar essas curvas a sistemas físicos.

```python
# A figura desta demonstração é estática; os três laboratórios acima mantêm os controles.
with plt.ioff():
    fig, ax = plt.subplots(figsize=(6, 5))
    t = np.linspace(0, 4*np.pi, 600)
    for sigma in [-.12, 0, .04]:
        z = np.exp((sigma+1j)*t)
        ax.plot(z.real, z.imag, label=f'σ = {sigma:g}')
    ax.set(xlabel='Re z', ylabel='Im z', aspect='equal', title='Magnitude variável, rotação constante')
    ax.legend(); ax.grid(alpha=.2); fig.tight_layout()
    # display estático para não criar um quarto painel de widgets.
    from IPython.display import display, Image
    from io import BytesIO
    buffer = BytesIO()
    fig.savefig(buffer, format='png', dpi=110)
    display(Image(data=buffer.getvalue()))
    plt.close(fig)
```
