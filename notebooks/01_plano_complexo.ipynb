# Uma coordenada a mais, uma geometria inteira

O ponto $3+4i$ fica três unidades à direita e quatro acima da origem. A seta revela a distância; as linhas pontilhadas revelam as duas coordenadas. [Leia o capítulo](../docs/01_plano_complexo.md).

```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / 'src' / 'complex_geometry.py').is_file()), None)
if ROOT is None:
    raise RuntimeError('Abra o notebook dentro da pasta do projeto.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
from src.complex_geometry import plot_complex

z = 3 + 4j
fig, ax = plt.subplots(figsize=(6, 6))
plot_complex(z, ax=ax)
ax.plot([z.real, z.real, 0], [0, z.imag, z.imag], ':', color='#b34a16')
ax.set_title('As projeções localizam o mesmo número')
plt.show()
print(f'Re(z) = {z.real:g}, Im(z) = {z.imag:g}, |z| = {abs(z):g}')
```

## Compare as direções

$1+i$, $-1+i$ e $-1-i$ apontam para quadrantes diferentes. O círculo revela o que permanece: todos têm módulo $\sqrt2$.

```python
fig, ax = plt.subplots(figsize=(6, 6))
for z, color in zip([1+1j, -1+1j, -1-1j], ['#1467ae', '#b34a16', '#087f8c']):
    ax.plot([0, z.real], [0, z.imag], 'o-', label=str(z), color=color)
t = np.linspace(0, 2*np.pi, 300)
ax.plot(np.sqrt(2)*np.cos(t), np.sqrt(2)*np.sin(t), ':', color='gray')
ax.axhline(0, color='gray', lw=.5); ax.axvline(0, color='gray', lw=.5)
ax.set(xlabel='Parte real', ylabel='Parte imaginária', aspect='equal')
ax.legend(); ax.grid(alpha=.2)
plt.show()
```
