# A álgebra move o plano

Multiplicar $1+i$ por $i$ leva o ponto a $-1+i$. Veja a rotação de 90° antes de confirmar a conta. [Leia o capítulo](../docs/05_operacoes_geometricas.md).

```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / 'src' / 'complex_geometry.py').is_file()), None)
if ROOT is None:
    raise RuntimeError('Abra o notebook dentro da pasta do projeto.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
z = 1+1j
result = 1j*z
fig, ax = plt.subplots(figsize=(6, 6))
for value, label, color in [(z, 'Antes: 1+i', '#1467ae'), (result, 'Depois: −1+i', '#b34a16')]:
    ax.plot([0, value.real], [0, value.imag], 'o-', color=color, label=label)
theta = np.linspace(np.pi/4, 3*np.pi/4, 90)
ax.plot(abs(z)*np.cos(theta), abs(z)*np.sin(theta), '--', color='#087f8c', label='Rotação de 90°')
ax.axhline(0, color='gray', lw=.5); ax.axvline(0, color='gray', lw=.5)
ax.set(xlabel='Parte real', ylabel='Parte imaginária', aspect='equal', xlim=(-2,2), ylim=(-2,2))
ax.legend(); ax.grid(alpha=.2); plt.show()
print(f'{z} × i = {result}; módulos: {abs(z):.4f} e {abs(result):.4f}')
```

## Quatro rotações fecham o percurso

Acompanhe $1\to i\to-1\to-i\to1$. Cada multiplicação acrescenta um quarto de volta; $i^4=1$ registra a volta completa.

```python
points = (1j)**np.arange(5)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(points.real, points.imag, 'o--', color='#1467ae')
for z, label in zip(points[:4], ['1', 'i', '−1', '−i']):
    ax.annotate(label, (z.real, z.imag), xytext=(8, 8), textcoords='offset points')
ax.set(xlabel='Parte real', ylabel='Parte imaginária', aspect='equal', xlim=(-1.4,1.4), ylim=(-1.4,1.4))
ax.grid(alpha=.2); plt.show()
print('Potências de i:', points)
```

Os segmentos ligam passos discretos, não a trajetória contínua da rotação.
