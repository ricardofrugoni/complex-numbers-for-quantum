## 3. Multiplicação de uma grade

Cada ponto passa pela mesma multiplicação. O módulo do multiplicador controla a escala e seu argumento controla a rotação. [OpenStax, seção 10.5](https://openstax.org/books/algebra-and-trigonometry-2e/pages/10-5-polar-form-of-complex-numbers).


```python
product_lab = multiplication()
plt.show()
```

**Exploração:** Compare multiplicadores com módulo 1 e ângulos 0°, 90° e 180°. Use módulo zero para observar o colapso da grade.


<a id="euler"></a>

## 4. Euler em 3D

A curva é $(\operatorname{Re}z(t),\operatorname{Im}z(t),t)$, com $z(t)=r e^{i(t+\phi)}$. O eixo $t$ é um parâmetro em radianos; **não é uma terceira parte do número complexo nem, por si só, tempo físico**. As projeções seguem diretamente da fórmula de Euler. [MIT, seção 1.6](https://ocw.mit.edu/courses/18-04-complex-variables-with-applications-spring-2018/6eae09885b0fab3f068e31752198abea_MIT18_04S18_topic1.pdf).


```python
euler_lab = euler_helix()
plt.show()
```

**Exploração:** Gire a câmera, mova o ponto no percurso e compare as componentes. Ao mudar apenas a câmera, os valores matemáticos devem permanecer iguais. Use “Restaurar vista” para recuperar a perspectiva inicial.


<a id="potencias"></a>

## 5. Potências no círculo

Para $z=e^{i\theta}$, $z^k=e^{ik\theta}$. O ângulo acumulado pode atravessar várias voltas; o argumento principal representa a mesma direção em um intervalo limitado. [OpenStax, seção 10.5](https://openstax.org/books/algebra-and-trigonometry-2e/pages/10-5-polar-form-of-complex-numbers).


```python
powers_lab = powers()
plt.show()
```

**Exploração:** Use θ=90° e n=4, depois θ=120° e n=3. Observe a repetição de pontos. Os segmentos ligam passos discretos, sem representar uma evolução contínua.


<a id="modulo"></a>

## 6. Superfície de módulo ao quadrado

A altura é $h=|a+bi|^2=a^2+b^2$. As curvas de nível são círculos com o mesmo módulo. Este é o gráfico de uma função real sobre o plano complexo. [OpenStax, seção 10.5](https://openstax.org/books/algebra-and-trigonometry-2e/pages/10-5-polar-form-of-complex-numbers).


```python
surface_lab = modulus_surface()
plt.show()
```

**Observe:** 1+i e −1+i têm a mesma altura, h=2. Ao mover o ponto para 2+2i, a altura chega a 8. O valor é válido para a função de módulo ao quadrado; interpretá-lo como probabilidade exigiria outro contexto e normalização.

<a id="amplitudes"></a>

## 7. Amplitudes normalizadas

Os controles geram $\alpha=\sqrt{p}\,e^{i\phi_\alpha}$ e $\beta=\sqrt{1-p}\,e^{i\phi_\beta}$, mantendo $|\alpha|^2+|\beta|^2=1$. O primeiro par de barras mostra P(0), P(1). O segundo usa $P(\pm)=|(\alpha\pm\beta)/\sqrt2|^2$, na base $|\pm\rangle=(|0\rangle\pm|1\rangle)/\sqrt2$. Cada par soma 1 separadamente: são duas escolhas de medição. A [ponte quântica](../docs/06_ponte_para_qubits.md#fase-relativa) explica essa janela para bases e interferência. [IBM Quantum Learning](https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/single-systems/quantum-information).


```python
amplitude_lab = amplitudes()
plt.show()
```

**Explore a fase relativa:** Use P(0)=0,5 e fase de α igual a 0°. Ao mover a fase de β de 0° até 180°, P(0) e P(1) permanecem em 0,5, enquanto P(+) vai de 1 a 0.

**Compare a fase global:** Compare os ajustes de fases (0°, 90°) e (45°, 135°). As duas setas giram, mas a diferença de fase e os dois pares de barras permanecem. Os estados intermediários, ao mover um controle de cada vez, têm outras diferenças de fase.

**Observe os extremos:** Use P(0)=0 ou P(0)=1. A fase da amplitude nula e a fase relativa ficam indefinidas; P(+)=P(−)=0,5.
