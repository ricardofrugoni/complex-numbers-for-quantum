# O círculo dentro da exponencial

$$e^{i\theta}=\cos\theta+i\sin\theta.$$

As duas componentes do mesmo ponto formam cosseno e seno ao longo do ângulo. [Leia o capítulo](../docs/04_forma_polar_euler.md).

```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / 'src' / 'complex_geometry.py').is_file()), None)
if ROOT is None:
    raise RuntimeError('Abra o notebook dentro da pasta do projeto.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
theta = np.linspace(0, 2*np.pi, 400)
z = np.exp(1j*theta)
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(z.real, z.imag, color='#1467ae')
for angle, label in zip([0, np.pi/2, np.pi, 3*np.pi/2], ['1','i','−1','−i']):
    point = np.exp(1j*angle)
    axes[0].plot(point.real, point.imag, 'o', color='#b34a16')
    axes[0].annotate(label, (point.real, point.imag), xytext=(7,7), textcoords='offset points')
axes[0].set(xlabel='Re z', ylabel='Im z', aspect='equal', xlim=(-1.4,1.4), ylim=(-1.4,1.4))
axes[1].plot(theta, z.real, label='cos θ · parte real')
axes[1].plot(theta, z.imag, '--', label='sen θ · parte imaginária')
axes[1].set(xlabel='θ (rad)', ylabel='Componente')
axes[1].legend()
for ax in axes: ax.grid(alpha=.2)
fig.tight_layout(); plt.show()
```

## Meia volta revela a identidade de Euler

No lado esquerdo do círculo, $e^{i\pi}=-1$, portanto $e^{i\pi}+1=0$. O computador trabalha com aproximações; o resíduo muito pequeno não contradiz a identidade exata.

```python
print('e^(iπ) + 1 =', np.exp(1j*np.pi)+1)
print('Próximo de zero:', np.isclose(np.exp(1j*np.pi)+1, 0))
```
